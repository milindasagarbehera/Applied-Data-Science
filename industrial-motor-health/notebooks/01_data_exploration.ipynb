## Step 5 — Summary Statistics

For each variable we compute:

### Mean

\[
\mu = \frac{1}{N} \sum_{i=1}^{N} x_i
\]

### Standard Deviation

\[
\sigma = \sqrt{ \frac{1}{N} \sum (x_i - \mu)^2 }
\]

These describe the **central tendency and variability** of the data.

```python
df.describe()
```

## Step 7 — Current Imbalance Feature

In a balanced 3‑phase motor:

\[
I_a = I_b = I_c
\]

Faults cause deviations.

We measure imbalance:

\[
I_{imbalance} = max(I_a,I_b,I_c) - min(I_a,I_b,I_c)
\]

Large values indicate **phase imbalance or winding faults**.

```python
df['current_imbalance'] = (
    df[['phase_current_a','phase_current_b','phase_current_c']].max(axis=1)
    - df[['phase_current_a','phase_current_b','phase_current_c']].min(axis=1)
)

df[['phase_current_a','phase_current_b','phase_current_c','current_imbalance']].head()
```

## Step 8 — Average Phase Current

The average current is

\[
I_{avg} = \frac{I_a + I_b + I_c}{3}
\]

This approximates the **overall motor load**.

```python
df['avg_current'] = (
    df['phase_current_a'] +
    df['phase_current_b'] +
    df['phase_current_c']
) / 3

df[['avg_current']].head()
```

## Step 9 — Histogram of Motor Temperature

## Step 10 — Correlation Matrix

Correlation measures linear relationships.

\[
r = \frac{Cov(X,Y)}{\sigma_X \sigma_Y}
\]

Where:

- Cov(X,Y) = covariance
- σ = standard deviation


```python
corr = df.corr(numeric_only=True)
corr
```
